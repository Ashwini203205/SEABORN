```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib .pyplot as plt
```

```python
sns.get_dataset_names()

df = sns.load_dataset("titanic")

df.head()
```

# Bar Plot
shows an estimate of a numeric value and its uncertainty across categories .
The vertical line on a seaborn column/bar plot is an error bar that indicates statistical uncertainty or data variability

```python
sns.barplot(x = df["sex"], y = df["fare"], palette =["yellow","blue"])
```

# Box Plot
 (useful for outliers)
 
--> to show data distribution, medians, quartiles, and outliers across categories.

Parts of a Box Plot : 

* Median (Middle Line): The exact middle value of your data.  (Half the numbers are higher, and half are lower)

* The Box (Middle 50%): Bottom edge (Q1): The 25th percentile. 25% of the data points are below this number.
                        Top edge (Q3)   : he 75th percentile. 75% of the data points are below this number.

* The Whiskers (Lines extending out): Show the normal range of the data up to the highest and lowest points (excluding strange numbers).

* Dots/Circles (Outliers): Strange data points that sit far away from the rest of the group.

```python
sns.boxplot(x = df["pclass"],y = df["fare"], palette =["red","green","blue"])    
```

```python
sns.boxplot(x = df["pclass"],y = df["fare"], palette =["red","green","blue"], hue = df["sex"])

#hue create two diff
```

# Violin Plot
-->  shows how numerical data is spread out across different groups. 

What Each Part Means : 

* Outer Curve (The Wide or Skinny Shape)  : Shows how often values happen (data density).
                                            A wide part means many data points have that value.
                                            A skinny part means very few data points have that value.

* The White Dot (in the middle):Shows the median (the exact middle number of your data).

* The Thick Black Bar (in the center):Shows the Interquartile Range (IQR), marking where the middle 50% of your data lives.

* The Thin Black Line (extending from the bar):Shows the rest of the main data range (high and low limits).

```python
sns.violinplot(x = df["pclass"],y = df["fare"], palette =["red","green"], hue=df["sex"])
```
