```python
sns.get_dataset_names()
```

```python
df = sns.load_dataset("titanic")
```

```python
df.head()
```

# Scatter Plot
labels will automatically comes in seaborn.

```python
sns.scatterplot(data=df, x="age", y="fare")      
```

```python
sns.scatterplot(data=df, x="age", y="fare", palette=["red","green","blue"], hue="class", size="pclass")     
```

# Histrogram / Distribution Plot
 For figure-level distribution plots,Pass a single numerical column or dataset to visualize how data is distributed across discrete intervals (bins).

```python
sns.histplot(df["age"], bins =15)       #bins : divinde in
```

```python
sns.histplot(df["age"], kde= True, bins =15)  
```

```python
sns.displot(df["age"], kde =True, bins =15)
```
