```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib .pyplot as plt
```

```python
sns.get_dataset_names()

df = sns.load_dataset("titanic")

df.head()
```

# Joint Plot
to view the bivariate relationship between two variables while simultaneously showing their univariate distributions on the outer edges.

```python
sns.jointplot( x = df["age"], y= df["fare"], hue = df["sex"], palette = ["red","green"], kind = "kde")   

# kde for circle curve 
```

```python
sns.jointplot(x = df["age"], y = df["fare"], hue = df["sex"], palette = ["red","green"], kind = "kde", fill=True)
 
  # fill for colur in circle
```

```python
sns.jointplot( x = df["age"], y= df["fare"], kind = "kde", fill = "True",  cmap = "YlGnBu") 

# if we want to compare between two category.
```

```python
sns.jointplot( x = df["age"], y= df["fare"], kind = "hex", cmap = "YlGnBu") 

# hex for hexa dimention shpe dark shows relative.(prsent)
```

# Heat Map

 function, which visualizes 2D tabular data as a color-encoded matrix. It is most commonly used for displaying correlation matrices and identifying patterns or density trends in datasets

```python
df.corr(numeric_only=True)  #shows all colums relationship between each others.
```

```python
sns.heatmap(df.corr(numeric_only=True))

# faint colurs or nera 1 are most corelated.
```

```python
sns.heatmap(df.corr(numeric_only=True), annot = True, cmap = "YlGnBu")

# annot = True for values inside box.
# cmap for change colur
```
